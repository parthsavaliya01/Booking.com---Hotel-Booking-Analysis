### Variables Description

11. children: Number of children included in the booking.
12. babies: Number of babies included in the booking.
13. meal: Type of meal booked (e.g., BB = Bed & Breakfast, HB = Half Board).
14. country: Country of origin of the guest.
15. market_segment: Market segment designation (e.g., Online, Offline, Corporate).
16. distribution_channel: Booking distribution channel (e.g., Direct, Online TA).
17. is_repeated_guest: Whether the booking is from a repeated guest (1 = Yes, 0 = No).
18. previous_cancellations: Number of previous bookings by the customer that were canceled.
19. previous_bookings_not_canceled: Number of previous bookings not canceled by the customer.
20. reserved_room_type: Code of the room type reserved.
21. assigned_room_type: Code of the room type assigned.
22. booking_changes: Number of changes made to the booking.
23. deposit_type: Type of deposit paid (e.g., No Deposit, Refundable, Non-Refundable).
24. agent: ID of the travel agency that made the booking.
25. company: ID of the company that made the booking.
26. days_in_waiting_list: Number of days the booking was on the waiting list.
27. customer_type: Type of customer (e.g., Transient, Group).
28. adr: Average Daily Rate per room per night.
29. required_car_parking_spaces: Number of car parking spaces required.
30. total_of_special_requests: Number of special requests made by the customer.
31. reservation_status: Status of the reservation (e.g., Check-Out, Canceled).
32. reservation_status_date: Date of the last reservation status update.

##### 1. Why did you pick the specific chart?

The histogram was chosen to visualize the distribution of the lead_time variable. This chart helps to:

* Show how frequently different ranges of lead times (the number of days between booking and arrival) occur.
* Identify any patterns, such as skewness or clustering in the data.
* Overlay the kernel density estimate (KDE) to provide a smoother view of the data's distribution.#

##### 2. What is/are the insight(s) found from the chart?

From the lead_time histogram:

* The data is likely right-skewed, with most bookings occurring at shorter lead times, but there are significant occurrences of very high lead times.
* A spike in bookings occurs in the range of 0 to 50 days, indicating that many customers tend to book within two months of their stay.
* The presence of outliers (very high lead times) suggests there are some bookings made far in advance, which could be attributed to specific events or group bookings.

##### 3. Will the gained insights help creating a positive business impact?

Yes, the insights can positively impact the business by:

* Optimizing marketing efforts: Understanding that most bookings happen within 0–50 days allows targeting customers with promotions or campaigns during this window.
